# Impact of Remote Work on Mental Health: Exploratory Data Analysis (EDA)

This notebook performs an exploratory data analysis on the dataset related to the impact of remote work on mental health. We will analyze the data to uncover key insights and patterns that might help understand how remote work affects mental well-being.

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.analysis import get_data_file_path
%matplotlib inline
```

## Load the Datasets
We begin by loading the dataset and inspecting its structure.

```python
# Load the dataset
csv_file_path = get_data_file_path()
df = pd.read_csv(csv_file_path)

# Print the first few rows
df.head(10)
```

## Data Overview
In this section, the basic information about the dataset was reviewed, including data types, missing values, and descriptive statistics.

```python
# Check for missing values and data types
print("----------- Dataset Info -----------")
df.info()

print("\n")

# Summary statistics
print("----------- Summary Statistics -----------")
print(df.describe())

print("\n")

# Checking missing values percentage per column
missing_values = df.isnull().sum() / len(df) * 100
print("----------- Missing Values Percentage per Column -----------")
missing_values
```

## Data Cleaning
Here the missing values, duplicates, and outliers were handled to prepare the data for analysis.


```python
# Drop rows with missing values
df_cleaned = df.dropna()

# Remove duplicates if any
df_cleaned = df_cleaned.drop_duplicates()

# You can choose to reset the index after cleaning
df_cleaned.reset_index(drop=True, inplace=True)

# Verify cleaning results
df_cleaned.info()
```

## Exploratory Data Analysis (EDA)
We perform exploratory data analysis by generating various plots to visualize the distribution of variables and relationships between them.

### Distribution of Key Variables
Next the visualization the distribution of the age, gender, job role and other.


```python
#plot the distribution of the age distribution
plt.figure(figsize=(5,4))
sns.histplot(df_cleaned['Age'], bins=10, kde=True)
plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# gender distribution
plt.figure(figsize=(5,4))
sns.countplot(x='Gender', data=df_cleaned)
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
# Job Roles distribution
plt.figure(figsize=(5, 4))
sns.countplot(data=df_cleaned, y='Job_Role', order=df_cleaned['Job_Role'].value_counts().index)
plt.title('Job Roles Distribution')
plt.xlabel('Count')
plt.ylabel('Job Role')
plt.show()
```

```python
mental_health_counts = df_cleaned['Mental_Health_Condition'].value_counts()
plt.figure(figsize=(5,4))
plt.pie(mental_health_counts, labels=mental_health_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette("Set2"))
plt.axis('equal')  
plt.title('Distribution of Mental Health Conditions', fontsize=16)
plt.show()
```

```python
# Satisfaction with Remote Work Distribution
satisfaction_counts = df_cleaned['Satisfaction_with_Remote_Work'].value_counts()
plt.figure(figsize=(4, 3))
plt.pie(satisfaction_counts, labels=satisfaction_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('Set2'))
plt.title('Satisfaction with Remote Work Distribution', fontsize=16)
plt.show()
```

```python
# Distribution of the social isolation
plt.figure(figsize=(5,4))
sns.histplot(df['Social_Isolation_Rating'], bins=10, kde=True)
plt.title('Distribution of Social Isolation Rating')
plt.xlabel('Social Isolation Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
# Work-Life Balance Rating by Work Location
sns.set(style="whitegrid")
plt.figure(figsize=(5,4))
sns.boxplot(x='Work_Location', y='Work_Life_Balance_Rating', data=df)
plt.title('Work-Life Balance Rating by Work Location', fontsize=16)
plt.xlabel('Work Location', fontsize=12)
plt.ylabel('Work-Life Balance Rating', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# stress level by gender
plt.figure(figsize=(5,4))
sns.countplot(x='Gender', hue='Stress_Level', data=df)
plt.title('Distribution of Stress Levels by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

### Correlation Analysis
Let's check the correlations between the numerical features in the dataset to understand potential relationships.


```python
# Correlation matrix
numeric_df = df_cleaned.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, 
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            square=True,
            cbar_kws={"shrink": .8})
plt.title('Correlation Matrix', fontsize=16)
plt.show()
```

### Relationship Between Remote Work and Mental Health
We will explore the relationship between remote work `Hours_Worked_Per_Week` and `Mental_Health_Condition`.


```python
# Scatter plot of work hours vs. mental health condition
df['Hours_Worked_Range'] = pd.cut(df['Hours_Worked_Per_Week'], bins=[0, 20, 40, 60, 80], labels=['0-20', '21-40', '41-60', '61-80'])
plt.figure(figsize=(6, 4))
sns.countplot(x='Hours_Worked_Range', hue='Mental_Health_Condition', data=df, palette='Set2')
plt.title('Mental Health Condition by Hours Worked Range', fontsize=16)
plt.xlabel('Hours Worked Per Week Range')
plt.ylabel('Count')
plt.legend(title='Mental Health Condition', fontsize='small', title_fontsize='medium', loc='lower right')
plt.show()
```

# Key Insights and Conclusions

## Key Insights

1. **Distribution of Key Variables**:
   - The dataset includes a diverse range of job roles and industries, with a noticeable proportion of employees experiencing mental health conditions.
   - The distribution of ages and genders suggests a varied workforce, which may influence mental health outcomes.

2. **Mental Health Condition Analysis**:
   - A significant number of respondents reported mental health conditions such as **anxiety**, **depression**, and **burnout**. This highlights the importance of mental health awareness in the workplace.

3. **Satisfaction with Remote Work**:
   - Insights from the analysis of satisfaction with remote work indicate a correlation between job roles and satisfaction levels. Some roles may require more social interaction, impacting their remote work satisfaction.
   - The distribution of satisfaction levels reveals that while some employees thrive in a remote work setting, others face challenges, suggesting the need for tailored approaches.

4. **Hours Worked Per Week and Mental Health**:
   - The analysis of hours worked per week against mental health conditions shows that longer hours may correlate with poorer mental health outcomes, underscoring the importance of maintaining a healthy work-life balance.



## Conclusion

- The findings from this analysis provide critical insights into the mental health landscape of the workforce.
- Employers should consider implementing supportive measures, such as mental health resources and flexible working arrangements, to enhance employee well-being.
- Tailored interventions may be necessary for different job roles and regions to effectively address mental health challenges.
- Future research could focus on longitudinal studies to assess how remote work dynamics evolve and their long-term impacts on employee mental health.

## Recommendations

- Promote mental health awareness programs within the organization.
- Conduct regular employee surveys to assess mental health and job satisfaction.
- Develop targeted strategies for employees in high-stress roles or regions with higher stress levels.
